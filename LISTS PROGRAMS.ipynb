```python
numbers=[]

for i in range(10):
    num=int(input("enter the value: "))
    numbers.append(num)

print("the final list=",numbers)

#sorted / sort: used to sort a list in ascending order

x=sorted(numbers)
print("ascending order=",x)

#to sort in descending order
numbers.sort(reverse=True)
print("in descending order the list will be ",numbers)

#to count number of elements in list
x=len(numbers)
print("total number of elements in the given list =",x)
```

```python
# loops
#to print the number of occurrences of a specified element in a list
numbers = [10, 20, 30, 20, 40, 20, 50]
target = 20

count = 0

for item in numbers:
    if item == target:
        count += 1  

print("the numbers of times",target,"been in the mentioned list= ",count)
```

```python
#Q:find ACCRONYM using the input from user
phrase=input("enter any phrase: ")
words=phrase.split()
accronym=""
for word in words:
    accronym=accronym+word[0]
    print("the accronym is",accronym.upper())
```

```python
#Q:abbrevation of months
months=["jan","feb","mar","api","may","jun","jul","aug","sep","oct","nov","dec"]
month_number=int(input("enter any number between 1-12:"))
if month_number>=1 and month_number<=12:
    print("month abbrevation:",months[month_number-1])
else:
    print("invalid number")
```

```python
#to identify and print common elements in 2 different lists
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common_elements = []

for item in list1:
    if item in list2:
        common_elements.append(item)

print("Common elements:", common_elements)
```
