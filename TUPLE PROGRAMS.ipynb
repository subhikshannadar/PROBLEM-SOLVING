```python
#TUPLE
numbers=(10,20,30,40,50,60,70,80,90)
print("TUPLE:",numbers)
print("4th element fron beginning:",numbers[3])
print("4th element from last:",numbers[-4])
```

```python
#to determine wheather numbers exist or not
numbers=(10,20,30,40,50,60,70,80,90)
element=int(input("enter the number:"))
if element in numbers:
    print("the entered element exists in tuple")
else:
    print("the given number does not exist in the tuple")
```

```python
#to change list to tuple
lists=[10,20,30,40]
tuples=tuple(lists)
print("list:",lists)
print("tuple",tuples)
```

```python
#index in tuple
numbers=(10,20,30,40)
elements=int(input("enter any number:"))
if elements in numbers:
    print("index of element is",numbers.index(elements))
else:
    print("it does not exist")
```

```python
#to replace last of tuples in a list
numbers=[(10,20,30),(40,50,60),(70,80,90)]
new_lists=[]
for tup in numbers:
    new_tuple=tup[:-1]+(100,)
    new_lists.append(new_tuple)

print("original list:",numbers)
print("new list:",new_lists)
```

```python
#to find maximum and minimum element in tuple
numbers=(10,20,30,40,50,60,70,80,90)
maximum=max(numbers)
print("the maximum number in the tuple=",maximum)
minimum=min(numbers)
print("the minimum number in the tuple=",minimum)
```

```python
#to swap first and last element in tuple
numbers=(10,20,30,40,50)
new_tuple=(numbers[-1],)+numbers[1:-1]+(numbers[0],)
print("original tuple=",numbers)
print("new tuple=",new_tuple)
```
