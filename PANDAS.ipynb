```python
#data cleaning and processing
import pandas as pn
data={
    "calories":[450,420,430,None,456],
    "duration":[50,60,70,75,None],
    "fluctuations":[20,None,None,10,20]
     
}
my_dict=pn.DataFrame(data)
print("original data : \n",my_dict)
print()

my_clean=my_dict.dropna()
print("clean data: \n",my_clean)
```

```python
import pandas as pn
data={
    "calories":[450,420,430,None,456],
    "duration":[50,60,70,75,None],
    "fluctuations":[20,None,None,10,20]
     
}
my_dict=pn.DataFrame(data)
print("original data : \n",my_dict.to_string(index=False))

print("duplicate rows: \n",my_dict[my_dict.duplicated()].to_string(index=False))

my_dict.drop_duplicates(subset=["calories"],keep="first",inplace=True)
print("data frames : \n" , my_dict.to_string(index=False))
```
