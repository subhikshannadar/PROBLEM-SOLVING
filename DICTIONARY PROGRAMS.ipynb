```python
#DICTIONARY PROGRAMME: 

#FIRST
#add a key and value to dict
my_dict={
    1:10,
    2:20}
print("original dict is : ",my_dict)
my_dict[3]=30
print("new dict is : ",my_dict)
```

```python
#SECOND
#concatenate the dict to create new one
dist1={1:10, 2:20, 3:30}
dist2={4:40 ,5:50, 6:60}
dist3={7:70, 8:80, 9:90}
new_dist={}
new_dist.update(dist1)
new_dist.update(dist2)
new_dist.update(dist3)
print("first dist is : ",dist1)
print("second dist is : ",dist2)
print("third dist is : ",dist3)
print("merged dist is : ",new_dist)
```

```python
#FORTH
#iterate to dict
my_dict={
    1:"raman",
    2:"rahul",
    3:"rohit",
    4:"palak",
    5:"rakshit",
    6:"abhijith",
}
for key in my_dict:
    print("keys are : ",key)
for value in my_dict.values():
    print("values are: ",value)
for key,value in my_dict.items():
    print("key:value is= ",key,value)
```

```python
#SIXTH
#sum of all value in dict
my_dict={"a":10, "b":20, "c":30, "d":40}
total=0
for value in my_dict.values():
    total=total+value
print("sum of all value is : ",total)
```

```python
#to sort dict by its values
my_dict={
    "rahul":94,
    "rohit":90,
    "subhikshan":98,
    "pall ":84
}
sorted_dict=dict(sorted(my_dict.items(),key=lambda item:item[1]))
print("dict sorted by values: ",sorted_dict)
sorted_dict=dict(sorted(my_dict.items(),key=lambda item:item[1],reverse=True))
print(sorted_dict)
```
